````python
# بجهز المودل ك templeate
# وهستخدمه لاحقا وادخل بيانات التدريب عليه

model = keras.Sequential(
        [
        keras.layers.Flatten(input_shape=(28,28)), # input Layer
        # كل ما زاد تعقيد الصور، بزيد عدد ال unit
       # 64  >> 128  >> 256
        keras.layers.Dense(64, activation="relu"), # hidden Layer 1
        keras.layers.Dense(10 , activation ="softmax") # output layers
        ]
)
````

````python
# بدي اجهز المودل ليصير جاهز للتدريب وعملية ال backprog.

model.compile(
    optimizer="adam", # كيف يعدل ال W
    loss=  "sparse_categorical_crossentropy", # كيف يحسب ال loss
    metrics =["accuracy"] # شو المقياس
)
````

````python
# EarlyStopping
callback = tf.keras.callbacks.EarlyStopping(patience=3,monitor="val_loss",restore_best_weights=True)
````

````python
history = model.fit(x_train,y_train ,epochs= 20, validation_data= (x_val,y_val),callbacks=callback )
````

````python

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
````

````python
from PIL import Image, ImageOps, ImageEnhance
import numpy as np
import matplotlib.pyplot as plt


from google.colab import files
import cv2  # للتعديل على الصور


uploaded = files.upload() #ارفع الملف
file_name = list(uploaded.keys())[0]

def prepare_and_predict(image_path):

    img = Image.open(image_path).convert('L') # نفتح الصورة و نحولها ل الرمادي

    # من المحاولات احافظ على الصورة في المنتصف
    width, height = img.size
    new_side = max(width, height)
    #لخلفية بيضا
    new_img = Image.new('L', (new_side, new_side), 255)
    new_img.paste(img, ((new_side - width)//2, (new_side - height)//2))

    #  الحجم ل  28*28
    img = new_img.resize((28, 28), Image.Resampling.LANCZOS)

    # هنا عكس الالوان ليكون الرقم ابيض ة الخلفية سوداء
    img_array = np.array(img)
    img_array = 255 - img_array

    #من المحاولات اوصل لافضل تحويل
    # لكن كان كله بنفس الاداء :(
    _, binary_img = cv2.threshold(img_array, 100, 255, cv2.THRESH_BINARY)

    # Normalization
    normalized_img = binary_img / 255.0
    # بدي اشوف الصورة الي انا ادخلتها و كيف تحولت عنده
    plt.figure(figsize=(6, 3))
    plt.subplot(1, 2, 1)
    plt.imshow(Image.open(image_path), cmap='gray')
    plt.title("Original")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(normalized_img, cmap='gray')
    plt.title("Processed for Model")
    plt.axis('off')
    plt.show()

    # المودل يختبر هنا ع لصورة الي دخلتها
    sample = normalized_img.reshape(1, 28, 28)
    prediction = model.predict(sample, verbose=0)
    predicted_label = np.argmax(prediction)

    print(f"Predicted Digit: {predicted_label}")
    print(f"Softmax Output Vector: {prediction[0].round(3)}")

    return predicted_label, prediction


prepare_and_predict(file_name)
````

````python
#Task 8


batch_sizes = [8, 32, 128]
histories = {}

for batch in batch_sizes:


    model = keras.Sequential([
            keras.layers.Flatten(input_shape=(28,28)),
            keras.layers.Dense(64, activation="relu"),
            keras.layers.Dropout(0.3),
            keras.layers.Dense(10, activation="softmax")
           ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

    # التدريب
    history = model.fit(x_train, y_train, epochs=20, batch_size=batch,
                        validation_data=(x_val, y_val), verbose=0)
    histories[batch] = history.history['loss']

# رسم النتائج في رسم بياني واحد للمقارنة
plt.figure(figsize=(10, 6))
for batch, loss_values in histories.items():
    plt.plot(loss_values, label=f'Batch Size {batch}')

plt.title('Effect of Batch Size on Gradient Noise (Loss Curves)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
````

````python
#task 9

activations = ['relu', 'tanh', 'gelu']
histories = {}

for act in activations:
    model = keras.Sequential([
        keras.layers.Flatten(input_shape=(28, 28)),
        keras.layers.Dense(64, activation=act),
        keras.layers.Dense(10, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

    # التدريب
    history = model.fit(x_train, y_train, epochs=20, batch_size=32,
                        validation_data=(x_val, y_val), verbose=0)
    histories[act] = history.history['val_loss']


plt.figure(figsize=(10, 6))
for act, val_loss in histories.items():
    plt.plot(val_loss, label=f'Activation: {act}')

plt.title('Activation Function Comparison (Validation Loss)')
plt.xlabel('Epochs')
plt.ylabel('Validation Loss')
plt.legend()
plt.show()
````
